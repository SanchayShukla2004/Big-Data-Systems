---
## Filter Operations

### Q1. Filter employees with salary greater than 50,000

```python
df.filter(col("salary") > 50000) \
  .select("employee_id", "name", "department", "salary") \
  .show(truncate=False)
```

### Q2. Filter employees whose age is above 30

```python
df.filter(col("age") > 30) \
  .select("employee_id", "name", "age", "department", "salary") \
  .show(truncate=False)
```

### Q3. Count total number of employees

```python
total_count = df.count()
print(f"Total number of employees: {total_count}")
```

### Q4. Show all employees from a specific city (Delhi)

```python
df.filter(col("city") == "Delhi") \
  .select("employee_id", "name", "role", "salary", "city") \
  .show(truncate=False)
```

### Q5. Select only name and salary columns

```python
df.select("name", "salary").show(20, truncate=False)
```

---
## GroupBy Operations

### Q6. Group by department and count employees in each department

```python
df.groupBy("department") \
  .agg(count("employee_id").alias("Employee_Count")) \
  .orderBy("Employee_Count", ascending=False) \
  .show(truncate=False)
```

### Q7. Find total salary per department

```python
df.groupBy("department") \
  .agg(spark_sum("salary").alias("Total_Salary")) \
  .orderBy("Total_Salary", ascending=False) \
  .show(truncate=False)
```

### Q8. Find average salary per department

```python
df.groupBy("department") \
  .agg(spark_round(avg("salary"), 2).alias("Average_Salary")) \
  .orderBy("Average_Salary", ascending=False) \
  .show(truncate=False)
```

### Q9. Group by city and find max salary

```python
df.groupBy("city") \
  .agg(spark_max("salary").alias("Max_Salary")) \
  .orderBy("Max_Salary", ascending=False) \
  .show(truncate=False)
```

### Q10. Group by department and gender and count employees

```python
df.groupBy("department", "gender") \
  .agg(count("employee_id").alias("Employee_Count")) \
  .orderBy("department", "gender") \
  .show(truncate=False)
```

---
## Multiple Aggregation

### Q11. Multiple aggregations on salary per department (sum, avg, max, min)

```python
df.groupBy("department") \
  .agg(
      spark_sum("salary").alias("Total_Salary"),
      spark_round(avg("salary"), 2).alias("Avg_Salary"),
      spark_max("salary").alias("Max_Salary"),
      spark_min("salary").alias("Min_Salary")
  ) \
  .orderBy("department") \
  .show(truncate=False)
```

### Q12. Rename all aggregated columns appropriately

```python
# All columns renamed using .alias() — professionally named
df.groupBy("department") \
  .agg(
      count("employee_id").alias("Headcount"),
      spark_sum("salary").alias("Sum_of_Salaries"),
      spark_round(avg("salary"), 2).alias("Average_Salary"),
      spark_max("salary").alias("Highest_Salary"),
      spark_min("salary").alias("Lowest_Salary")
  ) \
  .orderBy("department") \
  .show(truncate=False)
```

### Q13. Group by department and role — total salary, average salary, employee count

```python
df.groupBy("department", "role") \
  .agg(
      spark_sum("salary").alias("Total_Salary"),
      spark_round(avg("salary"), 2).alias("Average_Salary"),
      count("employee_id").alias("Employee_Count")
  ) \
  .orderBy("department", "Total_Salary", ascending=[True, False]) \
  .show(30, truncate=False)
```

### Q14. Count distinct roles in each department

```python
df.groupBy("department") \
  .agg(countDistinct("role").alias("Distinct_Roles")) \
  .orderBy("Distinct_Roles", ascending=False) \
  .show(truncate=False)
```

### Q15. Find the top 3 departments with the highest total salary

```python
df.groupBy("department") \
  .agg(spark_sum("salary").alias("Total_Salary")) \
  .orderBy(desc("Total_Salary")) \
  .limit(3) \
  .show(truncate=False)
```
