## Step 5: Check Data Types of the DataFrame
Use `printSchema()` to view column names and their data types.

```python
print("--- Schema (Column Names & Data Types) ---")
df.printSchema()

print("Column Names List:", df.columns)
print("Total Columns    :", len(df.columns))
print("Total Rows       :", df.count())
```

## Step 6: Check Summary Statistics
`describe()` gives count, mean, stddev, min, max for numeric columns.

```python
print("--- Basic Summary Statistics ---")
df.describe().show(truncate=False)

print("--- Extended Summary (with Percentiles) ---")
df.select("maths", "dbms", "python", "big_data", "ml").summary().show(truncate=False)
```

## Step 7: Select Columns
Three ways to select columns in PySpark.

```python
# Method 1: Select single column by name
print("-- Selecting 'name' column only --")
df.select("name").show(5)

# Method 2: Select multiple columns
print("-- Selecting 'name', 'department', 'python' --")
df.select("name", "department", "python").show(10, truncate=False)

# Method 3: Using col() object — recommended in production code
print("-- Using col() object to select 'name', 'maths', 'ml' --")
df.select(col("name"), col("maths"), col("ml")).show(5)
```

## Step 8: Adding a New Column
Use `withColumn()` to add computed columns.

```python
# Add 'total_marks' column: sum of all 5 subjects
df = df.withColumn(
    "total_marks",
    col("maths") + col("dbms") + col("python") + col("big_data") + col("ml")
)

# Add 'percentage' column: (total_marks / 500) * 100
df = df.withColumn(
    "percentage",
    (col("total_marks") / 500) * 100
)

print("-- DataFrame after adding 'total_marks' and 'percentage' --")
df.select("name", "maths", "dbms", "python", "big_data", "ml", "total_marks", "percentage") \
  .show(10, truncate=False)
```

## Step 9: Dropping a Column
Use `drop()` to remove one or more columns.

```python
# Drop single column
df_dropped = df.drop("semester")
print("-- After dropping 'semester' --")
print("Remaining columns:", df_dropped.columns)
df_dropped.show(3, truncate=False)

# Drop multiple columns at once
df_dropped_multi = df.drop("semester", "city")
print("\n-- After dropping 'semester' and 'city' --")
print("Remaining columns:", df_dropped_multi.columns)
```

## Step 10: Renaming a Column
Use `withColumnRenamed()` to rename columns.

```python
# Rename 'big_data' -> 'big_data_systems'
df = df.withColumnRenamed("big_data", "big_data_systems")

# Rename 'ml' -> 'machine_learning'
df = df.withColumnRenamed("ml", "machine_learning")

print("-- Updated column names after renaming --")
print(df.columns)

print("\n-- Final DataFrame Preview --")
df.show(10, truncate=False)
```
